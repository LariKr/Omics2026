```python
def print_n_to_m(n, m):
    while n <= m:
        print(n, end=" ")
        n = n + 1



n=1
m=15

print_n_to_m(n, m)

print("\n")

print_n_to_m(1, 5)

```

###Problem 6 in class

```python
def count_digits(n):
    count = 0
    while n > 0:
         n = n // 10
         count = count + 1
    return count

n = 984689
print(count_digits(n))
```
